## Exercise 11.1: A dictionary of words

Adapted from [Think Python 2e, Exercise 11.1](https://greenteapress.com/thinkpython2/html/thinkpython2012.html).

Complete `make_word_dict(filename="words.txt")`. Read the file and **return** a
dictionary whose keys are the words, with surrounding whitespace removed. Skip
blank lines. You may use `None` for every value. The file contains lowercase words.

Save the returned dictionary as `word_dict`. Display its number of keys and the
results of the membership checks below.

**Understanding:** Does `"hello" in word_dict` search keys or values? Why should
we remove the newline before using a word as a key?

````python
def make_word_dict(filename="words.txt"):
    word_dict = {}
    file = open(filename)
    for line in file:
        word = line.strip()
        if word != "":
            word_dict[word] = None
    return word_dict

word_dict = make_word_dict()
print(len(word_dict))
print("hello" in word_dict)
````

````python
# Run after completing make_word_dict().
word_dict = make_word_dict()
print("Number of words:", len(word_dict))
print("hello" in word_dict)       # Expected: True
print("hello\n" in word_dict)   # Expected: False
print("" in word_dict)            # Expected: False
````

**Your explanation:** Searching for something in a dictionary with 'in' always searches in keys, not values, so "hello" would be a key search. Removing the newline character removes it from being part of the key, leaving only the word (string).

## Exercise 11.4: Detecting duplicates with a dictionary

Adapted from [Think Python 2e, Exercise 11.4](https://greenteapress.com/thinkpython2/html/thinkpython2012.html).

Rewrite HW2's `has_duplicates(t)` using a dictionary to keep track of values
already encountered. Return `True` if any value repeats, and `False` otherwise.
Do not modify the input list, sort it, or replace the dictionary with a set.

As in HW2, inputs may mix integers and strings and will not contain nested lists.
All values used here are hashable and can be dictionary keys. An empty list has
no duplicates.

**Understanding:** What information does your dictionary represent as you traverse
the list, and why can you return as soon as you find a repeat? No timing experiment
or claim of a fixed speedup is required.

````python
def has_duplicates(t):
    bool_dict = {}
    for word in t:
        if word in bool_dict:
            return True
        bool_dict[word] = None
    return False
````

````python
print(has_duplicates(['a', 'b', 'b', 'c']))  # Expected: True
print(has_duplicates(['a', 'b', 'c', 'a']))  # Expected: True
print(has_duplicates(['a', 'b', 'c']))       # Expected: False
print(has_duplicates([]))                   # Expected: False
print(has_duplicates([1, '1', 2]))           # Expected: False

values = [3, 1, 3]
before = values.copy()
print(has_duplicates(values))               # Expected: True
print(values == before)                     # Expected: True
# Add one more test of your own and state its expected result.

print(has_duplicates(['x', 'y', 'z'])) # Expected: False
````

**Your explanation:** As I traverse the list, my dictionary's keys represent the values I have already encountered, so checking whether an item is a key tells me whether it has appeared before. I can return `True` as soon as I find a repeat because a single repeated value is enough to show the list has a duplicate, so the remaining items don't need to be checked.

## Exercise 11.5: Caesar-cipher rotate pairs

Adapted from [Think Python 2e, Exercise 11.5](https://greenteapress.com/thinkpython2/html/thinkpython2012.html).

A Caesar rotation shifts each letter forward through the alphabet, wrapping from
`z` to `a`. For example, `"cheer"` rotated by 7 becomes `"jolly"`.

1. Complete `rotate_word(word, shift)` for lowercase `a`–`z` words and integer
   shifts. Return a string. A shift of 26 leaves a word unchanged.
2. Complete `find_rotate_pairs(word_dict, min_length=5)`. Use the dictionary keys
   as the available words. Look for **distinct** words of at least `min_length`
   letters related by shifts 1 through 25.
3. Return a sorted list of tuples `(word1, word2, shift)`. Report each unordered
   pair once, with `word1 < word2` alphabetically. `shift` must rotate `word1`
   into `word2`. Do not also report the reverse pair or a word paired with itself.
4. Search the full word dictionary and display the **total number of pairs**.
   Then display only the **first 10 pairs from the sorted result**, one per line.
   Compute and return all pairs; limit only the displayed examples.

Hints: `ord()` converts a character to its numeric code and `chr()` converts a
code back to a character. Modulo 26 can handle wraparound. Generate rotations
and use dictionary membership instead of comparing every word with every other word.

**Understanding:** Why are shifts 1–25 sufficient for finding distinct rotate
pairs, and how does your output rule avoid counting a pair twice?

````python
def rotate_word(word, shift):
    string = ""
    for letter in word:
        offset = (ord(letter) - ord("a") + shift) % 26
        string += chr(ord("a") + offset)
    return string


def find_rotate_pairs(word_dict, min_length=5):
    pairs = []
    for word in word_dict:
        if len(word) >= min_length:
            for shift in range(1, 26):
                rotated = rotate_word(word, shift)
                if rotated in word_dict and word < rotated:
                    pairs.append((word, rotated, shift))
    return sorted(pairs)
````

````python
print(rotate_word("cheer", 7))  # Expected: jolly
print(rotate_word("zoo", 1))    # Expected: app
print(rotate_word("abc", 26))   # Expected: abc

small_words = {"cheer": None, "jolly": None, "melon": None, "a": None}
print(find_rotate_pairs(small_words))  # Expected: [('cheer', 'jolly', 7)]
# Add one more test of your own, including its expected result.
test_words = {"aptly": None, "timer": None, "arena": None, "a": None}
print(find_rotate_pairs(test_words))  # Expected: [('aptly', 'timer', 19)]
````

````python
# Find all pairs from word_dict; display the total count and the first 10 sorted pairs.

pairs = find_rotate_pairs(word_dict)
print(len(pairs))
for pair in pairs[:10]:
    print(pair)
````

**Your explanation:** Shifting 1-25 is sufficient because a shift of 26 would just loop back to the original character, and anything above 26 would be the same as 26 + any value from 1-25. My output avoids counting the same word twice by using the word < rotated line, since only one word of the two can come before the other.

Adapted from [Think Python 2e, Exercise 12.2](https://greenteapress.com/thinkpython2/html/thinkpython2013.html).

Anagrams use exactly the same letters with the same multiplicities. Complete
`group_anagrams(words)` for an iterable of distinct lowercase words. Return a
dictionary that maps a **tuple of sorted letters** to a list of its words.
For example, `"eat"` and `"tea"` share the key `('a', 'e', 't')`.

Use the keys of `word_dict` as input. Identify all groups containing **at least
six distinct words** and display the **total number of qualifying groups**.
Alphabetize the words within each qualifying group, then order the groups by
their first word. Display only the **first 10 groups** in that order, one per line.
Compute all groups before limiting the display. Groups with fewer than six words
should remain in the returned dictionary but need not be printed.

Example of the required group display:

```text
['abets', 'baste', 'bates', 'beast', 'beats', 'betas', 'tabes']
```

**Understanding:** Why can a tuple of letters be a dictionary key while the list
returned by `sorted(word)` cannot? Why must the key retain repeated letters, as
in `"abb"` versus `"ab"`?

````python
def group_anagrams(words):
    groups = {}
    for word in words:
        key = tuple(sorted(word))
        if key in groups:
            groups[key].append(word)
        else:
            groups[key] = [word]
    return(groups)
````

````python
small_groups = group_anagrams(["eat", "tea", "ate", "bat", "tab", "abb", "ab"])
print(sorted(small_groups[('a', 'e', 't')]))  # Expected: ['ate', 'eat', 'tea']
print(sorted(small_groups[('a', 'b', 't')]))  # Expected: ['bat', 'tab']
print(small_groups[('a', 'b', 'b')])          # Expected: ['abb']
print(small_groups[('a', 'b')])               # Expected: ['ab']
````

**Your explanation:** A tuple can be a dictionary key because it is immutable and therefore hashable, while the list returned by `sorted(word)` is mutable and unhashable, since its contents could change after being stored. We keep repeated letters because anagrams must match in letter counts: dropping repeats would give `"abb"` and `"ab"` the same key, `('a', 'b')`, and wrongly group them together.

### NP.2b: A Boolean mask for extreme values

Using the rounded array `c`, compute its observed mean and SD with `ddof=0`.
Select values whose **absolute distance from that mean is strictly greater than
three times that SD**. Use the observed quantities, not the generating parameters
100 and 15.

Display the selected values, their count, their proportion of all 1,000 values,
and the corresponding percentage. For example, 3 out of 1,000 would be a
proportion of `0.003`, or `0.3%`—not a proportion of `0.3`.

**Understanding:** Why does taking the mean of a Boolean mask give the proportion
selected? There is no requirement that this simulated sample have a particular
number of extreme values.

````python
# Compute the mask and display the values, count, proportion, and percentage.

extremes = c[((c - c.mean()) > (3 * c.std(ddof = 0))) | ((c.mean() - c) > (3 * c.std(ddof = 0)))]
print(extremes)
print(len(extremes))
print(len(extremes) / len(c))
print(len(extremes) / len(c) * 100, "%")
````

**Your explanation:** Taking the mean of a Boolean mask gives the proportion because all False values are 0 and all True values are 1, so the sum of the mask is the number of True values. Mathematically, taking the mean is equivalent to calculating the number of all True values divided by the number of all values, which is the proportion of True values in that sample.

### NP.8: Scale each column to 0–10

For the supplied `x`, linearly rescale **each column separately** so that its
minimum becomes 0 and its maximum becomes 10. Round to two decimal places and
store the result in `y`. Display `y` and its column minima and maxima.

These columns all have different minima and maxima. In the course environment,
the first two rows are:

```text
[[6.67 8.83 0.  ]
 [4.09 0.85 2.  ]]
```

**Understanding:** How is this transformation different from z-score
standardization? What would happen to your formula if a column were constant?
You do not need to implement a special rule for constant columns.

````python
# Supplied input; do not modify this cell. Same values as NP.7.
rng = np.random.default_rng(seed=100)
x = rng.integers(0, 100, size=30).reshape(10, 3)
````

````python
# Create y and display it, its column minima, and its column maxima.

col_mins = x.min(axis=0)
col_maxs = x.max(axis=0)

y = np.round((x - col_mins) / (col_maxs - col_mins) * 10, 2)

print(y)
print(y.min(axis=0))
print(y.max(axis=0))
````

**Your explanation:** This rescaling is different from z-score standardization because it uses each column's minimum and range to force the values into a fixed interval (0 to 10), rather than using the mean and SD to give a mean of 0 and SD of 1. If a column were constant, its maximum and minimum would be equal, so the formula would divide by zero (the range), producing `nan` values for that column.

### NP.9: Detect and replace NaN values

Make an independent copy of the supplied `x` named `x_filled`. Use a Boolean mask

**Understanding:** Why should you use `np.isnan(x)` instead of `x == np.nan`?
Replacing missing values with zero here is an array-manipulation exercise.
Give one reason it could distort an analysis if zero were not a justified value
for a missing measurement.

````python
# Supplied input; do not modify this cell.
rng = np.random.default_rng(seed=100)
x = rng.integers(0, 100, size=12).reshape(4, 3).astype(float)
row = np.array([1, 2, 3])
col = np.array([2, 1, 0])
x[row, col] = np.nan
````
